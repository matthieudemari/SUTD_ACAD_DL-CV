```python
class MyObject():
    def __init__(self):
        self.a = 10
    def __add__(self, other):
        return self.a + other.a
    def __getitem__(self, other):
        return "hello" + str(other)
    def __len__(self):
        return 1024
```

```python
x = MyObject()
print(x)
print(x.a)
```

```python
y = MyObject()
y.a = 12
print(y)
print(y.a)
```

```python
print(x + y)
```
